```python
import pandas as pd
import matplotlib.pyplot as plt

def analyze_cricket_reports(csv_filepath):
    # 1. Load Dataset
    df = pd.read_csv(csv_filepath)
    print("--- RAW DATA ---")
    print(df.head())
    print("\n" + "="*40 + "\n")

    # 2. Compute Additional Metrics
    # Calculate Strike Rate (Runs per 100 balls)
    df['Strike_Rate'] = (df['Runs'] / df['Balls']) * 100
    df['Strike_Rate'] = df['Strike_Rate'].fillna(0).round(2)

    # Calculate Economy Rate (Runs given / Overs) for bowlers
    df['Economy_Rate'] = df.apply(lambda row: round(row['Runs'] / row['Overs'], 2) if row['Overs'] > 0 else 0, axis=1)

    print("--- PROCESSED DATA WITH METRICS ---")
    print(df[['Player', 'Team', 'Runs', 'Balls', 'Strike_Rate', 'Wickets', 'Overs']])
    print("\n" + "="*40 + "\n")

    # 3. Top Run-Scorers Summary
    top_scorers = df.groupby(['Player', 'Team'])['Runs'].sum().reset_index()
    top_scorers = top_scorers.sort_values(by='Runs', ascending=False)
    print("--- TOP RUN SCORERS ---")
    print(top_scorers)

    # 4. Data Visualization - Runs Scored by Player
    plt.figure(figsize=(8, 5))
    plt.bar(top_scorers['Player'], top_scorers['Runs'], color='skyblue', edgecolor='black')
    plt.title('Total Runs Scored by Player')
    plt.xlabel('Player Name')
    plt.ylabel('Total Runs')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    
    # Save chart as image
    plt.tight_layout()
    plt.savefig('player_runs_summary.png')
    plt.show()
    print("\nVisualization saved as 'player_runs_summary.png'!")

if __name__ == "__main__":
    # Replace 'cricket_data.csv' with your actual CSV file name
    analyze_cricket_reports('cricket_data.csv')
```
